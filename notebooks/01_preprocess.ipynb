```python
import pandas as pd
import os

os.chdir("C:/Users/marco/job-search-engine")

df = pd.read_csv("data/raw_jobs.csv")

# Use the actual column names from this dataset
df = df[["title", "company_name", "description", "location"]].copy()
df = df.dropna(subset=["title", "description"])
df = df.drop_duplicates(subset=["title", "description"])

# Rename to standard names used throughout the project
df = df.rename(columns={
    "title": "job_title",
    "description": "job_description",
    "location": "job_location"
})

# Build search text field
df["search_text"] = df["job_title"] + ". " + df["job_description"].str[:500]

df.to_parquet("data/cleaned_jobs.parquet", index=False)
print(f"Saved {len(df)} jobs")
print(df.columns.tolist())
df.head(3)
```
